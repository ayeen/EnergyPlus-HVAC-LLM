# Slide 11: Performance Comparison
## Fine-Tuned vs. Pre-Trained Model Success Rates

```python
# Import Required Libraries
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches
```

```python
# Performance Comparison Data
models = ['Fine-Tuned\nMistral-7B', 'Pre-Trained\n(With Examples)', 'Pre-Trained\n(No Examples)']
success_rates = [100, 20, 0]

# Academic color palette — grayscale-friendly, print-safe
colors = ['#2c3e50', '#7f8c8d', '#bdc3c7']  # Dark, Medium, Light Gray
```

```python
# ── Academic Journal Style Configuration ──
import matplotlib as mpl

# Set Times New Roman (standard for engineering journals)
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'mathtext.fontset': 'stix',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 9,
})

# ── Create Figure (single-column journal width ~3.5in, double-column ~7in) ──
fig, ax = plt.subplots(figsize=(5.5, 4), dpi=300)
fig.patch.set_facecolor('white')

# ── Bar Chart ──
x = np.arange(len(models))
bar_width = 0.50
bars = ax.bar(x, success_rates, width=bar_width, color=colors,
              edgecolor='black', linewidth=0.6, zorder=3)

# ── Add hatch patterns for grayscale printing ──
hatches = ['///', '...', '']
for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)

# ── Value Labels ──
for bar, rate in zip(bars, success_rates):
    height = bar.get_height()
    offset = 2 if rate > 0 else 2
    ax.text(bar.get_x() + bar.get_width() / 2., height + offset,
            f'{rate}%', ha='center', va='bottom',
            fontsize=11, fontweight='normal', color='black')

# ── Title and Axis Labels ──
ax.set_ylabel('Success Rate (%)', fontsize=10, color='black')
ax.set_xlabel('Model Configuration', fontsize=10, color='black')

# ── Axis Configuration ──
ax.set_xticks(x)
ax.set_xticklabels(['Fine-Tuned\nLLM Model', 'Pre-Trained\n(With Examples)', 'Pre-Trained\n(No Examples)'],
                    fontsize=10, color='black')
ax.set_ylim(0, 115)
ax.set_yticks(range(0, 101, 20))

# ── Grid (subtle horizontal only) ──
ax.yaxis.grid(True, linestyle='--', alpha=0.4, linewidth=0.5, color='gray', zorder=0)
ax.xaxis.grid(False)

# ── Spines (academic style: left and bottom only) ──
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('black')
ax.spines['bottom'].set_color('black')
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)
ax.tick_params(axis='both', direction='in', length=4, width=0.8, colors='black')

# ── Legend ──
legend_patches = [
    mpatches.Patch(facecolor='#2c3e50', edgecolor='black', hatch='///', label='Fine-Tuned LLM Model'),
    mpatches.Patch(facecolor='#7f8c8d', edgecolor='black', hatch='...', label='Pre-Trained (With Examples)'),
    mpatches.Patch(facecolor='#bdc3c7', edgecolor='black', label='Pre-Trained (No Examples)')
]
ax.legend(handles=legend_patches, loc='upper right', fontsize=9,
          frameon=True, fancybox=False, edgecolor='black', framealpha=1.0)

plt.tight_layout()

# ── Save (journal-quality: 300 DPI minimum, EPS/PDF preferred) ──
plt.savefig('docs/figures/performance_comparison.png', dpi=300, bbox_inches='tight',
            facecolor='white', transparent=False)
plt.savefig('docs/figures/performance_comparison.pdf', dpi=300, bbox_inches='tight',
            facecolor='white')

plt.show()
print("Saved: docs/figures/performance_comparison.png (PNG)")
print("Saved: docs/figures/performance_comparison.pdf (PDF — recommended for journal submission)")
```
